Иногда одно и два можно сделать одновременно.

Мы будем использовать [линейную регрессию](https://en.wikipedia.org/wiki/Linear_reгрессия) для создания данных с известными **параметрами** (вещами, которые можно изучить с помощью модели), а затем воспользуемся PyTorch, чтобы посмотреть, сможем ли мы построить модель для оценки этих параметров с использованием [**градиентного спуска**](https://en.wikipedia.org/wiki/Gradient_descent).


````python
#Создать *известные* параметры
weight = 0.7
bias = 0.3

#Создать данные
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias

X[:10], y[:10]
````

Красивый! Теперь мы собираемся перейти к созданию модели, которая может изучить взаимосвязь между «X» (**функции**) и «y» (**метки**).


### Разделение данных на обучающие и тестовые наборы

Одним из наиболее важных шагов в проекте машинного обучения является создание набора для обучения и тестирования (и, при необходимости, набора для проверки).

Каждое разделение набора данных служит определенной цели:

| Сплит | Цель | Сумма общих данных | Как часто он используется? |
| ----- | ----- | ----- | ----- |
| **Тренировочный набор** | Модель учится на этих данных (например, на материалах курса, которые вы изучаете в течение семестра). | ~60-80% | Всегда |
| **Набор проверки** | Модель настраивается на основе этих данных (например, практический экзамен, который вы сдаете перед выпускным экзаменом). | ~10-20% | Часто, но не всегда |
| **Тестовый набор** | Модель оценивается на основе этих данных, чтобы проверить, чему она научилась (например, выпускной экзамен, который вы сдаете в конце семестра). | ~10-20% | Всегда |

На данный момент мы просто будем использовать набор для обучения и тестирования. Это означает, что у нас будет набор данных для нашей модели, на котором можно будет учиться и оценивать ее.

Мы можем создать их, разделив наши тензоры X и Y.

> **Примечание.** При работе с реальными данными этот шаг обычно выполняется в самом начале проекта (набор тестов всегда следует хранить отдельно от всех других данных). Мы хотим, чтобы наша модель обучалась на обучающих данных, а затем оценивала ее на тестовых данных, чтобы получить представление о том, насколько хорошо она **обобщает** на невидимые примеры.


````python
#Создать разделение поездов/тестов
train_split = int(0.8 * len(X)) #80% данных используется для обучающего набора, 20% для тестирования.
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

len(X_train), len(y_train), len(X_test), len(y_test)
````

Замечательно, у нас есть 40 образцов для обучения («X_train» и «y_train») и 10 образцов для тестирования («X_test» и «y_test»).

Модель, которую мы создаем, попытается изучить связь между «X_train» и «y_train», а затем мы оценим то, что она узнает на «X_test» и «y_test».

Но сейчас наши данные — это просто цифры на странице.

Давайте создадим функцию для ее визуализации.


````python
def plot_predictions(train_data=X_train,
                     train_labels=y_train,
                     test_data=X_test,
                     test_labels=y_test,
                     predictions=None):
  """
  Plots training data, test data and compares predictions.
  """
  plt.figure(figsize=(10, 7))

  #Отобразите данные обучения синим цветом
  plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")

  #Отобразите тестовые данные зеленым цветом
  plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")

  if predictions is not None:
    #Отобразите прогнозы красным цветом (прогнозы были сделаны на основе тестовых данных)
    plt.scatter(test_data, predictions, c="r", s=4, label="Predictions")

  #Показать легенду
  plt.legend(prop={"size": 14});
````

````python
plot_predictions()
````

## 2. Построение модели

Теперь у нас есть некоторые данные. Давайте построим модель, которая будет использовать синие точки для прогнозирования зеленых точек.

Мы собираемся прыгнуть прямо в воду.

Сначала напишем код, а потом все объясним.

Давайте воспроизведем стандартную модель линейной регрессии, используя чистый PyTorch.


````python
#Создайте класс модели линейной регрессии.
class LinearRegressionModel(nn.Module): #<- почти все в PyTorch представляет собой nn.Module (представьте себе это как блоки LEGO нейронной сети)
    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(1, #<- начните со случайных весов (это будет корректироваться по мере обучения модели)
                                                dtype=torch.float), #<- PyTorch по умолчанию поддерживает float32.
                                   requires_grad=True) #<- можем ли мы обновить это значение с помощью градиентного спуска?)

        self.bias = nn.Parameter(torch.randn(1, #<- начните со случайного смещения (оно будет корректироваться по мере обучения модели)
                                            dtype=torch.float), #<- PyTorch по умолчанию поддерживает float32.
                                requires_grad=True) #<- можем ли мы обновить это значение градиентным спуском?))

    #Вперед определяет вычисления в модели.
    def forward(self, x: torch.Tensor) -> torch.Tensor: #<- «x» — входные данные (например, функции обучения/тестирования)
        return self.weights * x + self.bias #<- это формула линейной регрессии (y = m*x + b)
````

Хорошо, выше много чего происходит, но давайте разберем это по частям.

### Основы построения моделей PyTorch

PyTorch имеет четыре (плюс-минус) основных модуля, которые вы можете использовать для создания практически любого типа нейронной сети, которую только можете себе представить.

Это [`torch.nn`](https://pytorch.org/docs/stable/nn.html), [`torch.optim`](https://pytorch.org/docs/stable/optim.html), [`torch.utils.data.Dataset`](https://pytorch.org/docs/stable/data.html#torch.utils.data.Dataset) и [`torch.utils.data.DataLoader`](https://pytorch.org/docs/stable/data.html). Сейчас мы сосредоточимся на первых двух, а к двум другим вернемся позже (хотя вы, возможно, догадаетесь, что они делают).

| Модуль PyTorch | Что он делает? |
| ----- | ----- |
| [`torch.nn`](https://pytorch.org/docs/stable/nn.html) | Содержит все строительные блоки для вычислительных графов (по сути, серии вычислений, выполняемых определенным образом). |
| [`torch.nn.Parameter`](https://pytorch.org/docs/stable/generated/torch.nn.parameter.Parameter.html#parameter) | Хранит тензоры, которые можно использовать с nn.Module. Если градиенты `requires_grad=True` (используемые для обновления параметров модели с помощью [**градиентного спуска**](https://ml-cheatsheet.readthedocs.io/en/latest/gradient_descent.html)) рассчитываются автоматически, это часто называют «автоградацией».  |
| [`torch.nn.Module`](https://pytorch.org/docs/stable/generated/torch.nn.Module.html#torch.nn.Module) | Базовый класс для всех модулей нейронных сетей, все строительные блоки нейронных сетей являются подклассами. Если вы создаете нейронную сеть в PyTorch, ваши модели должны создавать подклассы nn.Module. Требуется реализация метода `forward()`. |
| [`torch.optim`](https://pytorch.org/docs/stable/optim.html) | Содержит различные алгоритмы оптимизации (они сообщают параметрам модели, хранящимся в `nn.Parameter`, как лучше всего их изменить, чтобы улучшить градиентный спуск и, в свою очередь, уменьшить потери). |
| `def вперед()` | Все подклассы `nn.Module` требуют метода `forward()`, который определяет вычисления, которые будут происходить с данными, передаваемыми в конкретный `nn.Module` (например, формула линейной регрессии, приведенная выше). |

Если вышеизложенное звучит сложно, подумайте вот о чем: почти все в нейронной сети PyTorch происходит из «torch.nn»,
* `nn.Module` содержит более крупные строительные блоки (слои).
* `nn.Parameter` содержит меньшие параметры, такие как веса и смещения (объедините их вместе, чтобы создать `nn.Module`(s))
* `forward()` сообщает более крупным блокам, как выполнять вычисления на входах (тензорах, полных данных) внутри `nn.Module`(ов)
* torch.optim содержит методы оптимизации, позволяющие улучшить параметры внутри `nn.Parameter` для лучшего представления входных данных.

*Основные строительные блоки создания модели PyTorch путем создания подкласса nn.Module. Для объектов, являющихся подклассом nn.Module, должен быть определен метод forward().*

### Создание цикла оптимизации в PyTorch

The training loop involves the model going through the training data and learning the relationships between the `features` and `labels`.

Цикл тестирования включает в себя просмотр данных тестирования и оценку того, насколько хороши шаблоны, которые модель изучила на обучающих данных (модель никогда не видит данные тестирования во время обучения).

Each of these is called a "loop" because we want our model to look (loop through) at each sample in each dataset.


### Цикл обучения PyTorch
For the training loop, we'll build the following steps:

| Номер | Имя шага | Что он делает? | Пример кода |
| ----- | ----- | ----- | ----- |
| 1 | Пас вперед | The model goes through all of the training data once, performing its `forward()` function calculations. | `модель(x_train)` |
| 2 | Подсчитать убытки | The model's outputs (predictions) are compared to the ground truth and evaluated to see how wrong they are. | `loss = loss_fn(y_pred, y_train)` |
| 3 | Нулевые градиенты | The optimizers gradients are set to zero (they are accumulated by default) so they can be recalculated for the specific training step. | `optimizer.zero_grad()` |
| 4 | Выполнить обратное распространение ошибки | Computes the gradient of the loss with respect for every model parameter to be updated  (each parameter with `requires_grad=True`). This is known as **backpropagation**, hence "backwards".  | `loss.backward()` |
| 5 | Обновите оптимизатор (**градиентный спуск**) | Update the parameters with `requires_grad=True` with respect to the loss gradients in order to improve them. | `optimizer.step()` |

![цикл обучения pytorch с аннотациями](https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/01-pytorch-training-loop-annotated.png)

> **Примечание.** Вышеупомянутое — это лишь один пример того, как можно упорядочить или описать шаги. С опытом вы поймете, что создание обучающих циклов PyTorch может быть весьма гибким.
>
> And on the ordering of things, the above is a good default order but you may see slightly different orders. Некоторые практические правила:
> * Вычислите потери (`loss = ...`) *перед* выполнением обратного распространения ошибки (`loss.backward()`).
> * Нулевые градиенты (`optimizer.zero_grad()`) *перед* их шагом (`optimizer.step()`).
> * Шаг оптимизатора (`optimizer.step()`) *после* выполнения обратного распространения ошибки (`loss.backward()`).

Ресурсы, которые помогут понять, что происходит за кулисами обратного распространения ошибки и градиентного спуска, см. в разделе внеучебных программ.


### Цикл тестирования PyTorch

As for the testing loop (evaluating our model), the typical steps include:

| Номер | Имя шага | Что он делает? | Пример кода |
| ----- | ----- | ----- | ----- |
| 1 | Пас вперед | Модель проходит все обучающие данные один раз, выполняя вычисления с помощью функции «forward()». | `модель(x_test)` |
| 2 | Подсчитать убытки | Результаты модели (прогнозы) сравниваются с истиной и оцениваются, чтобы увидеть, насколько они ошибочны. | `loss = loss_fn(y_pred, y_test)` |
| 3 | Рассчитать показатели оценки (необязательно) | Alongisde the loss value you may want to calculate other evaluation metrics such as accuracy on the test set. | Пользовательские функции |

Обратите внимание, что цикл тестирования не содержит выполнения обратного распространения ошибки (`loss.backward()`) или пошагового выполнения оптимизатора (`optimizer.step()`), это связано с тем, что никакие параметры в модели не изменяются во время тестирования, они уже рассчитаны. Для тестирования нас интересуют только выходные данные прямого прохода через модель.

![цикл тестирования с аннотациями pytorch](https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/01-pytorch-testing-loop-annotated.png)

Давайте объединим все вышеперечисленное и обучим нашу модель в течение 100 **эпох** (прямой проход по данным) и будем оценивать ее каждые 10 эпох.


````python
torch.manual_seed(42)

#Установите количество эпох (сколько раз модель будет проходить обучение)
epochs = 100

#Создавайте пустые списки потерь для отслеживания значений
train_loss_values = []
test_loss_values = []
epoch_count = []

for epoch in range(epochs):
    ### Обучение

    #Переведите модель в режим обучения (это состояние модели по умолчанию).
    model_0.train()

    #1. Передача данных поезда вперед с использованием метода вперед() внутри
    y_pred = model_0(X_train)
    #print(y_pred)

    #Error 500 (Server Error)!!1500.That’s an error.There was an error. Please try again later.That’s all we know.
    loss = loss_fn(y_pred, y_train)

    #3. Нулевая градация оптимизатора
    optimizer.zero_grad()

    #4. Потеря назад
    loss.backward()

    #5. Продвигайте оптимизатор
    optimizer.step()

    ### Тестирование

    #Переведите модель в режим оценки
    model_0.eval()

    with torch.inference_mode():
      #1. Передача тестовых данных вперед
      test_pred = model_0(X_test)

      #2. Рассчитайте потери на тестовых данных.
      test_loss = loss_fn(test_pred, y_test.type(torch.float)) #прогнозы поступают в тип данных torch.float, поэтому сравнения необходимо проводить с тензорами того же типа.

      #Распечатайте, что происходит
      if epoch % 10 == 0:
            epoch_count.append(epoch)
            train_loss_values.append(loss.detach().numpy())
            test_loss_values.append(test_loss.detach().numpy())
            print(f"Epoch: {epoch} | MAE Train Loss: {loss} | MAE Test Loss: {test_loss} ")
````

Ох, вы бы посмотрели на это! Похоже, наши потери уменьшаются с каждой эпохой, давайте построим график, чтобы это выяснить.


````python
#Постройте кривые потерь
plt.plot(epoch_count, train_loss_values, label="Train loss")
plt.plot(epoch_count, test_loss_values, label="Test loss")
plt.title("Training and test loss curves")
plt.ylabel("Loss")
plt.xlabel("Epochs")
plt.legend()
````

## 4. Прогнозирование с помощью обученной модели PyTorch (вывод)

После обучения модели вы, вероятно, захотите делать с ее помощью прогнозы.

Мы уже видели это в приведенном выше коде обучения и тестирования. Действия вне цикла обучения/тестирования аналогичны.

При прогнозировании (также называемом выполнением вывода) с помощью модели PyTorch следует помнить три вещи:

1. Установите модель в режим оценки (`model.eval()`).
2. Сделайте прогнозы, используя контекстный менеджер режима вывода (`with torch.inference_mode(): ...`).
3. Все прогнозы следует делать с объектами на одном устройстве (например, данные и модель только на графическом процессоре или данные и модель только на процессоре).

Первые два пункта гарантируют, что все полезные вычисления и настройки, которые PyTorch использует «за кулисами» во время обучения, но не необходимы для вывода, отключены (это приводит к более быстрым вычислениям). ## 5. Сохранение и загрузка модели PyTorch

Если вы обучили модель PyTorch, скорее всего, вы захотите сохранить ее и куда-нибудь экспортировать.

Для сохранения и загрузки моделей в PyTorch вам следует знать три основных метода (все приведенные ниже сведения взяты из [Руководства по сохранению и загрузке моделей PyTorch] (https://pytorch.org/tutorials/beginner/saving_loading_models.html#saving-loading-model-for-inference)):

### Сохранение `state_dict()` модели PyTorch

[Рекомендуемый способ](https://pytorch.org/tutorials/beginner/saving_loading_models.html#saving-loading-model-for-inference) сохранения и загрузки модели для вывода (создания прогнозов) заключается в сохранении и загрузке метода `state_dict()` модели.

Давайте посмотрим, как это можно сделать в несколько шагов:

1. Мы создадим каталог для сохранения моделей под названием «models», используя модуль Python «pathlib».
2. Мы создадим путь к файлу для сохранения модели.
3. Мы вызовем `torch.save(obj, f)`, где `obj` — это `state_dict()` целевой модели, а `f` — это имя файла, в котором нужно сохранить модель.

> **Примечание.** Обычно сохраненные модели или объекты PyTorch заканчиваются на `.pt` или `.pth`, например `saved_model_01.pth`.


````python
from pathlib import Path

#1. Создайте каталог моделей.
MODEL_PATH = Path("models")
MODEL_PATH.mkdir(parents=True, exist_ok=True)

#2. Создайте путь для сохранения модели.
MODEL_NAME = "01_pytorch_workflow_model_0.pth"
MODEL_SAVE_PATH = MODEL_PATH / MODEL_NAME

#3. Сохраните словарь состояния модели.
print(f"Saving model to: {MODEL_SAVE_PATH}")
torch.save(obj=model_0.state_dict(), #только сохранение state_dict() сохраняет только изученные параметры модели
           f=MODEL_SAVE_PATH)
````

### Загрузка `state_dict()` сохраненной модели PyTorch

Поскольку теперь у нас есть сохраненная модель `state_dict()` в `models/01_pytorch_workflow_model_0.pth`, мы можем загрузить ее с помощью `torch.nn.Module.load_state_dict(torch.load(f))`, где `f` — это путь к файлу нашей сохраненной модели `state_dict()`.

Зачем вызывать torch.load() внутри torch.nn.Module.load_state_dict()?

Поскольку мы сохранили только `state_dict()` модели, который представляет собой словарь изученных параметров, а не *всю* модель, нам сначала нужно загрузить `state_dict()` с помощью `torch.load()`, а затем передать этот `state_dict()` в новый экземпляр нашей модели (который является подклассом `nn.Module`).

Почему бы не сохранить всю модель?

[Сохранение всей модели](https://pytorch.org/tutorials/beginner/saving_loading_models.html#save-load-entire-model), а не только `state_dict()`, более интуитивно понятно, однако, если процитировать документацию PyTorch (курсив мой):

> Недостаток этого подхода *(сохранения всей модели)* заключается в том, что сериализованные данные привязаны к конкретным классам и точной структуре каталогов, используемой при сохранении модели...
>
> Из-за этого ваш код может по-разному ломаться при использовании в других проектах или после рефакторинга.

Поэтому вместо этого мы используем гибкий метод сохранения и загрузки только `state_dict()`, который, по сути, снова представляет собой словарь параметров модели.

Давайте проверим это, создав еще один экземпляр LinearReprofitModel(), который является подклассом torch.nn.Module и, следовательно, будет иметь встроенный метод load_state_dict().


````python
#Создайте новый экземпляр нашей модели (он будет создан со случайными весами)
loaded_model_0 = LinearRegressionModel()

#Загрузите state_dict нашей сохраненной модели (это обновит новый экземпляр нашей модели с обученными весами)
loaded_model_0.load_state_dict(torch.load(f=MODEL_SAVE_PATH))
````

### 6.2 Построение линейной модели PyTorch

У нас есть некоторые данные, теперь пришло время создать модель.

Мы создадим тот же стиль модели, что и раньше, за исключением того, что на этот раз вместо определения параметров веса и смещения нашей модели вручную с помощью `nn.Parameter()` мы будем использовать [`nn.Linear(in_features, out_features)`](https://pytorch.org/docs/stable/generated/torch.nn.Linear.html), чтобы сделать это за нас.

Где in_features — это количество измерений, которые имеют ваши входные данные, а out_features — это количество измерений, в которые вы хотите, чтобы они выводились.

В нашем случае оба они равны «1», поскольку наши данные имеют «1» входной признак («X») для каждой метки («y»).

! [сравнение модели линейной регрессии nn.Parameter и модели линейной регрессии nn.
*Создание модели линейной регрессии с использованием nn.Parameter вместо использования nn.Linear. Существует еще множество примеров того, как модуль «torch.nn» имеет встроенные вычисления, включая множество популярных и полезных слоев нейронных сетей.*


````python
#Подкласс nn.Module для создания нашей модели
class LinearRegressionModelV2(nn.Module):
    def __init__(self):
        super().__init__()
        #Используйте nn.Linear() для создания параметров модели.
        self.linear_layer = nn.Linear(in_features=1,
                                      out_features=1)

    #Определите прямое вычисление (входные данные x проходят через nn.Linear())
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear_layer(x)

#Установите начальное значение вручную при создании модели (это не всегда необходимо, но используется в демонстрационных целях, попробуйте закомментировать и посмотреть, что произойдет)
torch.manual_seed(42)
model_1 = LinearRegressionModelV2()
model_1, model_1.state_dict()
````

````python
torch.manual_seed(42)

#Установите количество эпох
epochs = 1000

#Поместите данные на доступное устройство
#Без этого произойдет ошибка (не все модели/данные на устройстве)
X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)

for epoch in range(epochs):
    ### Обучение
    model_1.train() #режим поезда включен по умолчанию после постройки

    #1. Пас вперед
    y_pred = model_1(X_train)

    #2. Рассчитайте убытки
    loss = loss_fn(y_pred, y_train)

    #3. Оптимизатор нулевой градации
    optimizer.zero_grad()

    #4. Проигрыш назад
    loss.backward()

    #5. Шаг оптимизатора
    optimizer.step()

    ### Тестирование
    model_1.eval() #перевести модель в режим оценки для тестирования (вывода)
    #1. Пас вперед
    with torch.inference_mode():
        test_pred = model_1(X_test)

        #2. Подсчитайте убытки
        test_loss = loss_fn(test_pred, y_test)

    if epoch % 100 == 0:
        print(f"Epoch: {epoch} | Train loss: {loss} | Test loss: {test_loss}")
````

> **Примечание.** Из-за случайного характера машинного обучения вы, скорее всего, получите немного разные результаты (разные значения потерь и прогнозов) в зависимости от того, была ли ваша модель обучена на ЦП или на графическом процессоре. Это верно, даже если вы используете одно и то же случайное начальное число на обоих устройствах. Если разница большая, возможно, стоит поискать ошибки, однако, если она небольшая (в идеале так оно и есть), ее можно игнорировать.
